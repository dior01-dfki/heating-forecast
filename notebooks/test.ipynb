```python
import pandas as pd
#import darts 
from darts import TimeSeries
from darts.models import LinearRegressionModel
from darts.metrics import mape
from darts.utils.missing_values import extract_subseries, fill_missing_values
```

```python
df = pd.read_csv('../data/test_data.csv')
df['ts'] = pd.to_datetime(df['ts'])
```

```python
df.isna().sum()
```

```python
df['ts'] = df['ts'].dt.tz_convert(None) 
```

```python
df.drop(columns=['building_id'], inplace=True)
```

```python
value_cols = list(df.columns)
value_cols.remove('ts')
value_cols.remove('room_id')
```

```python
darts_ts = TimeSeries.from_group_dataframe(df, time_col='ts', value_cols=value_cols, freq='H', group_cols=['room_id'])
```

```python
darts_ts[0].components
```

```python
df['room_id'].nunique()
```

```python
extracted_ts = [extract_subseries(ts,min_gap_size=4, mode="any") for ts in darts_ts]
```

```python
len(extracted_ts)
```

```python
extracted_ts
```

```python
from darts.models import LinearRegressionModel

model = LinearRegressionModel(lags=24, output_chunk_length=12)
```

```python
for ts in extracted_ts:
    try:
        model.fit(ts)
        print(f"Fitted model for time series with room_id: {ts['room_id'].values()[0]}")
    except Exception as e:
        print(f"Error fitting model for time series: {e}")
```
