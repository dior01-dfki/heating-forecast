```python
import logging
import sys

# Get the root logger
root_logger = logging.getLogger()
root_logger.setLevel(logging.INFO)

# Remove existing handlers (avoid duplicates in Jupyter)
if root_logger.hasHandlers():
    root_logger.handlers.clear()

# Add a stream handler to show logs in notebook/stdout
stream_handler = logging.StreamHandler(sys.stdout)
stream_handler.setLevel(logging.INFO)
formatter = logging.Formatter("%(asctime)s - %(levelname)s - %(name)s - %(message)s", "%H:%M:%S")
stream_handler.setFormatter(formatter)

root_logger.addHandler(stream_handler)

print("✅ Logger configured for notebook output")
```

```python
import pandas as pd

from forcateri import TimeSeries


if __name__ == "__main__":
    steps, deltas = 3, 2
    dti = pd.date_range(start="2000-01-01", freq="D", periods=steps)
    tdi = pd.timedelta_range(start=0, freq="D", periods=deltas)
    mi = pd.MultiIndex.from_product([tdi, dti])

    df = pd.DataFrame(
        data=range(steps * deltas),
        index=mi,
        columns=pd.MultiIndex.from_product(
            [["some_feature"], ["value"]], names=TimeSeries.COL_INDEX_NAMES
        ),
    )
    ts = TimeSeries(df)
```
